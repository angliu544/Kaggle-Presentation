```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("online food delivery dataset.csv")
```

```python
dataset.info()
dataset.head()
```

```python
dataset_clean = dataset.drop(columns=["latitude", 
                                      "longitude", 
                                      "Pin code", 
                                      "Output", 
                                      "Feedback", 
                                      "Unnamed: 13"])
```

```python
dataset_clean.info()
dataset_clean.head()
```

```python
dataset_clean.describe()
```

```python
columns = [
    "Gender",
    "Marital Status",
    "Occupation",
    "Monthly Income",
    "Educational Qualifications",
    "Customer Type"
]

for column in columns:
    print(
        dataset_clean[column]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
        .astype(str) + "%"
    )
    print()
```

```python
plt.figure(figsize=(6, 4))
ax = sns.boxplot(data=dataset_clean, x="Customer Type", y="Age", palette=["#ffb9b9", "#8dbccd", "#BBFCBB"])
ax.set_xticklabels(["Frequent", "Regular", "New"])
plt.title("Age Spread By Order Frequency")
plt.xlabel("Order Frequency")
plt.ylabel("Age")
plt.show()
```
