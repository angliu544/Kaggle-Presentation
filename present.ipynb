```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("online food delivery dataset.csv")
```

```python
dataset.info()
dataset.head()
```

```python
dataset_clean = dataset.drop(columns=["latitude", 
                                      "longitude", 
                                      "Pin code", 
                                      "Output", 
                                      "Feedback", 
                                      "Unnamed: 13"])
```

```python
dataset_clean.info()
dataset_clean.head()
```

```python
dataset_clean.describe()
```

```python
columns = [
    "Gender",
    "Marital Status",
    "Occupation",
    "Monthly Income",
    "Educational Qualifications",
    "Customer Type"
]

for column in columns:
    print(
        dataset_clean[column]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
        .astype(str) + "%"
    )
    print()
```

```python
# Box Plot
plt.figure(figsize=(6, 4))
ax = sns.boxplot(data=dataset_clean, x="Customer Type", y="Age", palette=["#ffb9b9", "#8dbccd", "#BBFCBB"])
ax.set_xticklabels(["Frequent", "Regular", "New"])
plt.title("Age Spread By Order Frequency")
plt.xlabel("Order Frequency")
plt.ylabel("Age")
plt.show()
```

```python
# Pie Chart
counts = dataset_clean["Occupation"].value_counts()
plt.pie(counts, labels=counts.index, colors=["#ffb9b9", "#8dbccd", "#BBFCBB", "#f7f1af"], autopct="%1.1f%%")
plt.show()
```

```python
# Bar Graph
plt.figure(figsize=(7, 4))
ax = sns.countplot(data=dataset_clean, x="Occupation", palette=["#ffb9b9", "#8dbccd", "#BBFCBB", "#f7f1af"])
ax.set_xticks([0, 1, 2, 3])
ax.set_xticklabels(["Student","Employee","Self Employeed","House Wife"])
plt.title("Occupations vs. Frequency")
plt.xlabel("Occupation")
plt.ylabel("Frequency")
plt.show()
```
